```python
import numpy as np
```

```python
def read_matrix_from_binary_file(filename, is_vec = False):
    with open(filename, "rb") as f:
        # Read the number of rows and columns
        rows = int.from_bytes(f.read(4), byteorder='little', signed=True)

        if is_vec:
            # If it's a vector, read only one dimension
            cols = 1
        else:
            # Read the number of columns
            cols = int.from_bytes(f.read(4), byteorder='little', signed=True)

        print(rows, cols)
        
        # Read the matrix data
        data = np.fromfile(f, dtype=np.float64)
        return data.reshape((rows, cols))
```

```python
nrows = 934
ncols = 612

# Example usage
filename = f"inputs/input_matrix_{nrows:04d}_{ncols:04d}.bin"
matrix = read_matrix_from_binary_file(filename)
print(matrix[:5, :5])  # Print the first 5x5 submatrix
```

```python
# Read correct outputs
filename = f"correct_output/u_{nrows:04d}_{ncols:04d}.bin"
corr_U = read_matrix_from_binary_file(filename)

filename = f"correct_output/v_{nrows:04d}_{ncols:04d}.bin"
corr_V = read_matrix_from_binary_file(filename)

# Read the correct singular values
filename = f"correct_output/singular_{nrows:04d}_{ncols:04d}.bin"
corr_S = read_matrix_from_binary_file(filename, is_vec=True)
```

```python
print(corr_S[:5])  # Print the first 5 singular values
```
